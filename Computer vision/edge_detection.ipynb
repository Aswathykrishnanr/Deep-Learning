```python
import cv2
import os
```

```python
image = cv2.imread("open-road.jpg", cv2.COLOR_BGR2GRAY)
```

```python
#gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
blur = cv2.GaussianBlur(image, (5, 5), 0)
```

```python
edges = cv2.Canny(blur, 50, 150)
```

```python
cv2.imshow("road_edges.png", edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
