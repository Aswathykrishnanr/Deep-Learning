```python
import ultralytics
import cv2
```

```python
model=ultralytics.YOLO('yolov8n.pt')
```

```python
video=cv2.VideoCapture(0)
while True:
    succ,img =video.read()
    result=model(img)
    for i in result:
        i.save(filename='result1.jpg')
        img=cv2.imread('result1.jpg')
        cv2.imshow('Video',img)
    if cv2.waitKey(1) & 0XFF == ord('q'):
        break
video.release()
cv2.destroyAllWindows()
```
