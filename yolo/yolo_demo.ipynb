```python
import ultralytics
```

```python
model=ultralytics.YOLO('yolov8n.pt')
```

```python
import cv2

results = model(['street.jpg', 'car.jpg'])

for i in results:
    i.save(filename='result.jpg')
    img = cv2.imread('result.jpg')

    cv2.imshow("Result", img) 
    cv2.waitKey(1000)
    cv2.destroyAllWindows()
```
